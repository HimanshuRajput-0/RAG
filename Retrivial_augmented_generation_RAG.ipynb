# Document conversion tools for RAG.

The effectiveness of any RAG system heavily depends on how well is its knowledge source is and to create an effective knowledge source we need to process and convert them into searchable vectors.

Poorly extracted or malformed content can leads to semantic drift in the vector space and hence can impact the quality of the RAG system.

A typical processing pipeline for RAG involves several processes.

1. Document Parsing

2. Context Extraction

3. Cleaning

4. Chunking

5. Vectorization

Each stage presents unique challenges for handling complex layouts in PDF, DOCs, PPTs to preserve semantic relation for long terms.
